### 2. I Selected only the columns that will be useful for the analysis

```python
df = df[['brand', 'categories', 'dateAdded' , 'colors', 'name', 'features', 'imageURLs',
       'prices.amountMax', 'prices.color', 'prices.currency', 'reviews']]

df.head()
```

### 3. Cleaning `'Brand'` column

```python
df['brand'] = df['brand'].str.lower().str.strip().str.replace('@', 'a',  regex=True)


brands = ['aerosoles', 'kenneth cole', 'ellie', 'toms', 'ugg', 'pleaser', 'michael kors', 'ralph lauren',
          'puma', 'walking cradles', 'vionic', 'vince', 'victoria', 'valentino', 'adidas', 'bcbg', 'sperry',
          'spring step', 'nina' , 'moda'
         ]

for brand in  brands :
    filt = df['brand'].str.contains(brand).fillna(False)
    df.loc[filt, 'brand'] = brand
```

**- This column had a lot of the same brands named differently, I joined them to their correct name**

```python
non_brands = ['famous name brand', 'generic', 'generic surplus', 'non-branded',
             'not applicable', 'not rated', 'lucky  brand', 'lucky brand', 'very fine dance shoes'
             ]
filt = df['brand'].isin(non_brands)
df.loc[filt, 'brand'] = 'unbranded'
df['brand'].fillna('unbranded', inplace = True)
```

**- I labeled missing valued and wrong brand names as `'Unbranded'` .**

```python
ls = df['brand'].value_counts()[df['brand'].value_counts() < 20]
ls = list(ls.index)
filt = df['brand'].isin(ls)
df.loc[filt, 'brand'] = 'small brands'


df['brand'] = df['brand'].str.capitalize()

df['brand'].value_counts()
```

**- I grouped all the small brands that had less than 30 products on the market to one category `'Small Brand'`**

### 4. Cleaning `'features'` Column

```python
df['features'][100]
```

```python
def get_gender(text):
    text = text.lower()
    
    keys = re.search(r'"key":"gender","value":\["([\w\.-]+)"\]', text)
    if keys == None:
        return np.nan
    else :
        return keys.group(1)
```

```python
gender = df['features'].astype('str').apply(get_gender)
gender.unique()
```

```python
filt = gender.isin(['men', 'boys'])
df[filt].head()
```

```python
df = df[~filt]
```

**- Here I extracted the feature `'gender'` and it turned out that some `'men'` and `'boys'` shoes slipped into the dateset, so I removed them.**

```python
def get_color(text):
    text = text.lower()
    keys = re.search(r'"key":"color","value":\["([\w\.-]+)"\]', text)
    if keys == None:
        return np.nan
    else :
        return keys.group(1)
```

```python
df['features'].replace('colour', 'color', inplace = True)
colors = df['features'].astype('str').apply(get_color)
colors.value_counts()
```

```python
df['colors'] = df['colors'].fillna(colors)
```

```python
b = pct["colors"]
b = round(b, 2)
a = 100 * df["colors"].isna().sum()/df.shape[0]
a = round(a, 2)

print(f'Missing Values percentage is reduced from {b} to {a} \n')
```

**- Here I extracted the feature `'color'` then I used it to fill the missing values in the column `'colors'` .**

```python
b = 100 * df["colors"].isna().sum()/df.shape[0]

filt = df['prices.color'].str.lower().str.contains('woman|women|inches|medium|small|large')
df.loc[filt.fillna(False), ['prices.color']] = np.nan
```

```python
df['colors'] = df['colors'].fillna(df['prices.color'])
a = 100 * df["colors"].isna().sum()/df.shape[0]

print(f'Missing Values percentage is reduced from {b} to {a}\n')
```

```python
df['colors'] = df['colors'].str.lower().str.strip()
df.drop(columns = ['prices.color'], inplace = True)
```

```python
df.drop('features', axis = 1, inplace = True)
```

**- I did the same with `'prices.color'` and used it to fill the missing values in `'color'` column**

**- However I've noticed that `'prices.color'` wrong values of sizes so I'd to remove them first**

**- Then I deleted the column `'prices.color'`**

**- I removed the very expensive outliers of the `'Small Brand'`**

### 6. Cleaning `'categories'` Column

```python
not_shoes = 'boy|skirt|accessories|jewelry|bag|luggage'
df.drop(df[df['categories'].str.lower().str.contains(not_shoes)].index, inplace = True)
```

- First I removed all producted that are not shoes

- Then, I'll group the shoes into 8 categories
    - Boots
    - Heels
    - Sandals
    - Sports Shoes
    - Causual Shoes
    - Kids Shoes
    - Designer Shoes
    - Others


```python
boots = 'boots|boot|bootie|booties|calf|leather|work|oxfords|pu|paten'
df.loc[df['categories'].str.lower().str.contains(boots), 'categories'] = 'Boots'

heels = 'pump|heels|stiletto|wedge|platform'
df.loc[df['categories'].str.lower().str.contains(heels), 'categories'] = 'Heels'

sandals = 'sandal|gladiator|flip|flop|slipper|slide|flat'
df.loc[df['categories'].str.lower().str.contains(sandals), 'categories'] = 'Sandals'


sports = 'walking|cleats|access|athletic|basketball|golf|sport|run|running|trainin|gym|fitness|climb|soccer|outdoor|hiking|hike|skecher|balance|fila|asics|Easy Spirit|trainer|adidas|puma|reebok|nike'
df.loc[df['categories'].str.lower().str.contains(sports), 'categories'] = 'Sports'

causual = 'comfort|toning|patio|sinclair|casual|contemporary|sneaker|slip-on|clog|round|toe|converse|canvas|mule|moccasin|flat|Bahama|espadrille|loafer|skechers'
df.loc[df['categories'].str.lower().str.contains(causual), 'categories'] = 'Causual'

kids = 'kid|girl|child|school|uniform|toy'
df.loc[df['categories'].str.lower().str.contains(kids), 'categories'] = 'Kids'

design = 'design'
df.loc[df['categories'].str.lower().str.contains(design), 'categories'] = 'Design'

women_shoes = 'women|woman'
df.loc[df['categories'].str.lower().str.contains(women_shoes), 'categories'] = 'Other Women Shoes'
```

```python
df['categories'].value_counts()
```

```python
df['categories'] = df['categories'].astype('category')
```

### 5. Cleaning `'prices'` Column

```python
df.rename({'prices.amountMax' : 'prices'}, axis = 1, inplace = True)

to_usd = {'USD' : 1, 'CAD' : 0.75, 'EUR' : 1.09, 'AUD' : 0.67, 'GBP' : 1.27}

for curr, fct in to_usd.items() :
    filt = (df['prices.currency'] == curr)
    df.loc[filt, 'prices'] = fct * df.loc[filt, 'prices']
    
df.loc[df['prices.currency'] != 'USD',['prices', 'prices.currency']].head()
```

```python
df.drop('prices.currency', axis = 1, inplace = True)
```

**- I Converted all prices to USD**

```python
df[['prices']].describe()
```

```python
filt = (df['brand'].isin(['Unbranded', 'Small brands'])) & (df['prices'] > 1500)
df.loc[filt, ['brand', 'categories', 'prices']]
```

```python
df.drop(df[filt].index, inplace = True)

df[['prices']].describe()
```

### 6. `'reviews'` column

```python
df['reviews'][0]
```

```python
def get_rating(txt):
    rgx = r'"rating":(\d+\.\d+)'
    mtch = re.search(rgx, txt)
    if mtch == None :
        return np.nan
    else:
        return mtch.group(1)
    
df['rating'] = df['reviews'].fillna('').apply(get_rating)
df['rating'] = pd.to_numeric(df['rating'])
```

```python
100*df['rating'].isna().sum()/df.shape[0]
```

```python
df.drop('reviews', axis = 1, inplace = True)

df.head()
```

**- I extracted the rating of the product from the `'reviews'` column, however almost %90 of the colomn is missing so it's not really reliable for the analysis**

### 7. `'dateAdded'` Column

```python
df['year'] = pd.to_datetime(df['dateAdded']).dt.year
df.drop('dateAdded', axis = 1,inplace = True)

df.head()
```

**- Extracted the year and droped the column**

### 8. `'imageURLs'` Column

```python
cleaned = df['imageURLs'].str.split(',').dropna().apply(lambda x : x[0])
df.loc[cleaned.index, 'imageURLs'] = cleaned
```

```python
filt = df['imageURLs'].duplicated(keep = 'first')
df.loc[filt, 'imageURLs'] = np.nan
```

**- Removed duplicates**
